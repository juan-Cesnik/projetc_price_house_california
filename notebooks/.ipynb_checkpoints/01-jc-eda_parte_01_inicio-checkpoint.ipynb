```python
fig, axs = plt.subplots(3,3, figsize=(10,5))

for ax, coluna in zip(axs.flatten(), df.columns):
    sns.boxplot(data=df, x=coluna, ax=ax, showmeans = True)

plt.tight_layout()

plt.show
```

```python
matriz = np.triu(df.select_dtypes("number").corr())

fig, ax = plt.subplots()

sns.heatmap(
    df.select_dtypes("number").corr(),
    mask=matriz,
    annot=True,
    fmt=".2f",
    ax=ax,
    cmap = PALETTE
)

plt.show()
```

```python
# Se viu que tem um a correlação forte é a renda e isso não é bom indicio e tambem muito valores muito baixos porque no seu modelo elel só vai ter uma unica coisa importante e isso não é bom
```

```python
#Quando analisamos o heatmap você percebe correlação entre variaveis diferente e conforme uma aumenta a outra tambem aumenta e com isso tambem podemos criar novas colunas como:
```

```python
"""
Algumas variáveis novas:

* Criar classes em median_income;
* quartos por domicilio;
* pessoas por domicilio;
* quartos por comodo;
"""
```

```python
#Criando uma nova coluna categórica
```

```python
#vou usar o cat para indicar que é uma variavel categorica
```

```python
df["median_income_cat"] = pd.cut( #estamos criando colunas categoricas e cada numero representa uma categorica e ela vai ser uma categoria ordinal
    df["median_income"],
    bins=[0, 1.5, 3, 4.5, 6, np.inf], #esse np.inf, estou indicando para deixar em aberto porque se futuramento adicionar mais dados que por exemplo pode ser maior do que eu passei.
    labels=[1,2,3,4,5],
)

df.info()
```
